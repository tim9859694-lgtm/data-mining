```python
import pandas as pd
import numpy as np
import statsmodels.formula.api as smf
from pathlib import Path
import os

# 1. 경로 설정 및 자동 디렉토리 생성 (오류 방지)
DATA_DIR = Path('/content/type3')
os.makedirs(DATA_DIR, exist_ok=True)  # 폴더가 없으면 자동으로 생성해줍니다.

# [주의] 이 코드를 실행하기 전, 코랩 좌측 파일 탭의 'type3' 폴더 안에
# 'churn_logit1.csv' 파일이 실제로 업로드되어 있어야 합니다!
csv_path = DATA_DIR / 'churn_logit1.csv'

if not csv_path.exists():
    print(f"❌ [경고] {csv_path} 파일이 존재하지 않습니다.")
    print("👉 코랩 좌측 [파일] 메뉴 -> 'type3' 폴더 생성 -> 'churn_logit1.csv' 파일을 업로드해 주세요.")
else:
    # 2. 데이터 불러오기
    df = pd.read_csv(csv_path)

    print("=== 데이터 미리보기 ===")
    print(df.head())
    print("\n=== 기본 정보 ===")
    print(df.info())
    print("-" * 50)

    # 3. 로지스틱 회귀 적합
    # 종속변수(churn)와 독립변수들이 데이터프레임에 올바른 문자열로 존재하는지 확인하세요.
    model = smf.logit('churn ~ age + usage_hour + complaint_cnt', data=df).fit()

    print("\n=== 로지스틱 회귀 요약 ===")
    print(model.summary())
    print("-" * 50)

    # 4. 계수표 정리
    result_table = pd.DataFrame({
        'coef': model.params,
        'p_value': model.pvalues,
        'odds_ratio': np.exp(model.params)
    })

    print("\n=== 계수 / p-value / 오즈비 ===")
    print(result_table)
    print("-" * 50)

    # 5. 95% 신뢰구간과 오즈비 기준 정리
    conf = model.conf_int()
    conf.columns = ['2.5%', '97.5%']
    conf['OR_2.5%'] = np.exp(conf['2.5%'])
    conf['OR_97.5%'] = np.exp(conf['97.5%'])

    print("\n=== 계수 신뢰구간 및 오즈비 신뢰구간 ===")
    print(conf)
    print("-" * 50)

    # 6. 예측확률 및 분류
    df['pred_prob'] = model.predict(df)
    df['pred_class'] = (df['pred_prob'] >= 0.5).astype(int)

    print("\n=== 예측확률 상위 5개 ===")
    print(df[['churn', 'pred_prob', 'pred_class']].head())
    print("-" * 50)

    # 7. 해석용 출력
    print("\n=== 해석 가이드 ===")
    for var in ['age', 'usage_hour', 'complaint_cnt']:
        coef = model.params[var]
        pval = model.pvalues[var]
        or_val = np.exp(coef)
        direction = '증가' if coef > 0 else '감소'
        print(f"▶️ {var}: 계수={coef:.4f}, p-value={pval:.6f}, 오즈비={or_val:.4f}")
        print(f"   -> {var}가 1단위 증가할 때 이탈 Odds(승산)는 약 {or_val:.4f}배가 되며, 방향은 {direction}입니다.\n")
```

```python
import pandas as pd
import numpy as np
import statsmodels.formula.api as smf
from sklearn.metrics import accuracy_score
from pathlib import Path
import os

# 1. 경로 설정 및 자동 디렉토리 생성 (오류 방지)
DATA_DIR = Path('/content/type3')
os.makedirs(DATA_DIR, exist_ok=True)  # 폴더가 없으면 자동으로 생성

# [주의] 코랩 좌측 파일 탭의 'type3' 폴더 안에 'loan_default_logit2.csv' 파일이 있어야 합니다.
csv_path = DATA_DIR / 'loan_default_logit2.csv'

if not csv_path.exists():
    print(f"❌ [경고] {csv_path} 파일이 존재하지 않습니다.")
    print("👉 코랩 좌측 [파일] 메뉴 -> 'type3' 폴더 생성 -> 'loan_default_logit2.csv' 파일을 업로드해 주세요.")
else:
    # 2. 데이터 불러오기
    df = pd.read_csv(csv_path)

    print("=== 데이터 미리보기 ===")
    print(df.head())
    print("\n=== 기본 정보 ===")
    print(df.info())
    print("-" * 50)

    # [안전 장치] 분석에 사용할 변수에 결측치(NaN)가 있으면 제거하여
    # 이후 accuracy_score 계산 시 데이터 개수 불일치 오류를 방지합니다.
    df = df.dropna(subset=['default', 'income', 'debt_ratio', 'late_cnt']).reset_index(drop=True)

    # 3. 로지스틱 회귀 적합
    model = smf.logit('default ~ income + debt_ratio + late_cnt', data=df).fit()

    print("\n=== 로지스틱 회귀 요약 ===")
    print(model.summary())
    print("-" * 50)

    # 4. 계수 / p-value / 오즈비
    result_table = pd.DataFrame({
        'coef': model.params,
        'p_value': model.pvalues,
        'odds_ratio': np.exp(model.params)
    })
    print("\n=== 계수 / p-value / 오즈비 ===")
    print(result_table)
    print("-" * 50)

    # 5. 로그우도와 residual deviance
    llf = model.llf
    residual_deviance = -2 * llf

    print("\n=== 적합도 지표 ===")
    print(f"log-likelihood (llf): {llf:.4f}")
    print(f"residual deviance: {residual_deviance:.4f}")
    print("-" * 50)

    # 6. 예측확률, 예측분류
    df['pred_prob'] = model.predict(df)
    df['pred_class'] = (df['pred_prob'] >= 0.5).astype(int)

    # 7. 성능지표 (위에서 결측치를 밀어주었기 때문에 무조건 안전하게 작동합니다)
    acc = accuracy_score(df['default'], df['pred_class'])
    err = 1 - acc

    print("\n=== 분류 성능 ===")
    print(f"accuracy     : {acc:.4f}")
    print(f"error rate   : {err:.4f}")
    print("-" * 50)
```
