```python
from pathlib import Path
import pandas as pd
import statsmodels.formula.api as smf
import os

# 1. 경로 설정 및 파일 확인
# 코랩의 현재 작업 디렉토리 또는 드라이브 마운트 경로를 확인하세요.
DATA_DIR = Path('/content/type3')
file_path = DATA_DIR / 'retail_sales_reg1.csv'

# 파일이 실제로 존재하는지 체크
if not file_path.exists():
    print(f"❌ [오류] 지정한 경로에 파일이 없습니다: {file_path}")
    print("💡 [해결 방법] 코랩 왼쪽 메뉴의 폴더 아이콘(📁)을 눌러 'type3' 폴더와 'retail_sales_reg1.csv' 파일이 올바르게 업로드되었는지 확인해 주세요.")
    print(f"현재 /content 폴더 내부 목록: {os.listdir('/content') if os.path.exists('/content') else '없음'}")
else:
    # 2. 데이터 불러오기 (파일이 존재할 때만 실행)
    df = pd.read_csv(file_path)

    print('[데이터 상위 5행]')
    print(df.head(), '\n')

    print('[기본 정보]')
    print(df.info(), '\n')

    # 3. 회귀모형 적합
    model = smf.ols('sales ~ ad_cost + staff + event_cnt', data=df).fit()

    # 4. 결과 요약
    print('[회귀 요약 결과]')
    print(model.summary())

    # 5. 핵심 결과 추출
    print('\n[회귀계수]')
    print(model.params)

    print('\n[p-value]')
    print(model.pvalues)

    print('\n[R-squared]')
    print(model.rsquared)

    # 6. 광고비 10단위 증가 효과
    beta_ad = model.params['ad_cost']
    print('\n광고비 10단위 증가 시 예상 매출 변화:', beta_ad * 10)

    # 7. 유의한 변수 목록
    sig_vars = model.pvalues[model.pvalues < 0.05].index.tolist()
    print('\n유의한 변수:', sig_vars)
```

```python
from google.colab import drive
from pathlib import Path
import pandas as pd
import numpy as np
import statsmodels.formula.api as smf
from sklearn.metrics import mean_squared_error
import os

# 1. 경로 설정 및 폴더 자동 생성
DATA_DIR = Path('/content/type3')
if not DATA_DIR.exists():
    DATA_DIR.mkdir(parents=True, exist_ok=True)
    print("📢 'type3' 폴더가 없어서 새로 생성했습니다. 코랩 왼쪽 📁 메뉴에서 해당 폴더에 'weather_reg2.csv'를 업로드해 주세요!")

file_path = DATA_DIR / 'weather_reg2.csv'

# 2. 파일 존재 여부 최종 확인 후 실행
if not file_path.exists():
    print(f"❌ [오류] 아직 폴더에 파일이 올라오지 않았습니다. 경로를 다시 확인하세요: {file_path}")
else:
    # 3. 데이터 불러오기
    df = pd.read_csv(file_path)
    print("✅ 데이터를 성공적으로 불러왔습니다!\n")

    print('[데이터 상위 5행]')
    print(df.head(), '\n')

    print('[기본 정보]')
    print(df.info(), '\n')

    # 4. 회귀모형 적합
    model = smf.ols('temperature ~ solar + wind + o3', data=df).fit()

    # 5. 결과 요약
    print('[회귀 요약 결과]')
    print(model.summary())

    # 6. 핵심 계수와 p-value
    print('\n[회귀계수]')
    print(model.params)

    print('\n[p-value]')
    print(model.pvalues)

    print('\n[R-squared]')
    print(model.rsquared)

    # 7. 특정 조건 예측
    new_df = pd.DataFrame({
        'solar': [100],
        'wind': [5],
        'o3': [30]
    })

    pred = model.predict(new_df)
    print('\n[예측값]')
    print(pred)

    # 8. 신뢰구간 / 예측 요약표
    pred_frame = model.get_prediction(new_df).summary_frame(alpha=0.05)
    print('\n[예측 요약표]')
    print(pred_frame)

    # 9. 오차지표 계산
    y_true = df['temperature']
    y_pred = model.predict(df)

    mse = mean_squared_error(y_true, y_pred)
    rmse = np.sqrt(mse)

    print('\n[MSE / RMSE]')
    print('MSE :', mse)
    print('RMSE:', rmse)
```
